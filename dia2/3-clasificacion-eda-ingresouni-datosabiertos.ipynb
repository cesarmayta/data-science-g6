```python
df_cleaned = df.drop(columns=['IDHASH', 'COLEGIO', 'COLEGIO_PROV','COLEGIO_DIST','COLEGIO_PAIS','DOMICILIO_DEPA','DOMICILIO_PROV','DOMICILIO_DIST','NACIMIENTO_PAIS','NACIMIENTO_DEPA','NACIMIENTO_PROV','NACIMIENTO_DIST','CALIF_FINAL','MODALIDAD'])
df_cleaned
```

```python
df_cleaned['egreso_postulacion'] = df_cleaned['ANIO_POSTULA'] - df_cleaned['COLEGIO_ANIO_EGRESO']
df_cleaned['edad'] = df_cleaned['ANIO_POSTULA'] - df_cleaned['ANIO_NACIMIENTO']
df_cleaned
```

```python
df_cleaned = df_cleaned.drop(columns=['COLEGIO_ANIO_EGRESO', 'ANIO_POSTULA', 'CICLO_POSTULA','ANIO_NACIMIENTO'])
df_cleaned
```

```python
df_cleaned.isnull().sum()
```

```python
df_cleaned = df_cleaned.dropna()
df_cleaned
```
